````python

import sympy as sp
from sympy import symbols, Eq, Function, exp, sin, cos, log, sqrt, integrate, dsolve, diff, simplify, solve, series, summation, factorial, limit, Abs

# ---------- Setup ----------
x = sp.symbols('x')
y = sp.Function('y')

print("04a - Q1 (a) to (d): general solutions\n")

# (1)(a) dy/dx = x^2 / y
ode_a = Eq(sp.diff(y(x), x), x**2 / y(x))
print("a:", dsolve(ode_a))

# (1)(b)
ode_b = Eq(sp.diff(y(x), x) - exp(x)/y(x)**2, 0)
print("b:", dsolve(ode_b))

# (1)(c)
ode_c = Eq(sp.diff(y(x), x) + y(x)**3*sp.cos(x), 0)
print("c:", dsolve(ode_c))

# (1)(d)
ode_d = Eq(exp(3*x)*sp.diff(y(x), x) - exp(y(x)), 0)
print("d:", dsolve(ode_d))
````

````python
# ---------- Q2 particular ----------
print("\nQ2 particular solutions")
ode2a = Eq(sp.diff(y(x), x) - x**2*y(x)**4, 0)
print("a:", dsolve(ode2a, ics={y(1):1}))

# ---------- Q3 integrate twice ----------
print("\nQ3 integrate twice")
C1, C2 = sp.symbols('C1 C2')
ypp = sqrt(2*x - 1)
yp = sp.integrate(ypp, x) + C1
y_expr = sp.integrate(yp, x) + C2
eqs = [Eq(yp.subs(x,1), 4/3), Eq(y_expr.subs(x,1), 16/15)]
consts = sp.solve(eqs, (C1, C2))
print("a:", simplify(y_expr.subs(consts)))

# ---------- Q4 ----------
sp.init_printing()

x, y = sp.symbols('x y')
z = sp.Function('z')
A = sp.Function('A')
C = sp.Function('C')

# Q4(a), (b), (c): Mixed partial derivatives

# (a) ∂²z/∂x∂y = x*y²
f_a = x * y**2
# integrate w.r.t x
dz_dy_a = sp.integrate(f_a, x) + sp.Function('g1')(y)
# integrate w.r.t y
z_a = sp.integrate(dz_dy_a, y) + sp.Function('h1')(x)
print("Q4(a) z(x, y) =")
sp.pprint(sp.simplify(z_a))

# (b) ∂²z/∂x∂y = e^y cos x
f_b = sp.exp(y)*sp.cos(x)
dz_dy_b = sp.integrate(f_b, x) + sp.Function('g2')(y)
z_b = sp.integrate(dz_dy_b, y) + sp.Function('h2')(x)
print("\nQ4(b) z(x, y) =")
sp.pprint(sp.simplify(z_b))

# (c) ∂²z/∂x∂y = y/x + 2
f_c = y/x + 2
dz_dy_c = sp.integrate(f_c, x) + sp.Function('g3')(y)
z_c = sp.integrate(dz_dy_c, y) + sp.Function('h3')(x)
print("\nQ4(c) z(x, y) =")
sp.pprint(sp.simplify(z_c))
````

````python
# Tutorial 4(b) – 01. Training Accuracy

# dA/dx = A / x^2
ode_A = sp.Eq(sp.diff(A(x), x), A(x)/x**2)

# (a) General solution
sol_A_general = sp.dsolve(ode_A)
print("General solution for A(x):")
sp.pprint(sol_A_general)

# (b) Particular solution: A(1) = 1
sol_A_particular = sp.dsolve(ode_A, ics={A(1):1})
print("\nParticular solution (A(1)=1):")
sp.pprint(sol_A_particular)

# Cell 4 – 02. Operational Cost

# ∂²C/∂x∂y = sin(x) sin(y)
f_C = sp.sin(x)*sp.sin(y)

# integrate w.r.t x
dC_dy = sp.integrate(f_C, x) + sp.Function('g')(y)

# integrate w.r.t y
C_general = sp.integrate(dC_dy, y) + sp.Function('h')(x)
print("General solution for C(x, y):")
sp.pprint(sp.simplify(C_general))

# Apply given conditions:
# ∂C/∂y = -2 sin(y) when x=0, and C=0 when y=0
# From ∂C/∂y at x=0, we can determine g(y)
dC_dy_expr = sp.diff(C_general, y)
g_y = sp.Function('g')(y)

# Evaluate ∂C/∂y at x=0
dC_dy_x0 = sp.simplify(dC_dy_expr.subs(x, 0))
print("\n∂C/∂y at x=0 =", dC_dy_x0)

# Set condition: ∂C/∂y = -2 sin(y)
# => g'(y) = -2 sin(y)  => integrate
g_sol = sp.integrate(-2*sp.sin(y), y)
print("\ng(y) =", g_sol, "+ constant (absorbed later)")

# Substitute g(y)
dC_dy_new = sp.integrate(f_C, x) + g_sol
C_particular = sp.integrate(dC_dy_new, y) + sp.Function('h')(x)

# Apply C(x,0)=0
C_y0 = C_particular.subs(y,0)
h_sol = -C_y0
C_final = sp.simplify(C_particular.subs(sp.Function('h')(x), h_sol))

print("\nParticular C(x, y) satisfying conditions:")
sp.pprint(C_final)

````
