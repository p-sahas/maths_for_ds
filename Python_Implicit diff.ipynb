Q1 (a)

```python
import sympy as sp

# Define symbols
x = sp.Symbol('x')
y = sp.Function('y')(x)   # y is a function of x

# Define the equation
expr = x**2 + y**3 - 4   # x^2 + y^3 - 4 = 0

# Differentiate implicitly wrt x
dexpr_dx = sp.diff(expr, x)

# Simplify result
dexpr_dx_simplified = sp.simplify(dexpr_dx)

print("Derivative equation:")
sp.pprint(dexpr_dx_simplified)

# Solve for dy/dx
dy_dx = sp.solve(dexpr_dx_simplified, sp.Derivative(y, x))
print("\ndy/dx =")
sp.pprint(dy_dx[0])
```
