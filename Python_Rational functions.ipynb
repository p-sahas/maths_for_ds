Q2 (a)

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define variable
x = sp.Symbol('x')

# Define function
f1 = (x**2 + 1) / (x + 1)

# Derivative and critical points
df1 = sp.diff(f1, x)
critical_points1 = sp.solve(sp.simplify(df1), x)
second_derivative1 = sp.diff(f1, x, 2)

print("f1(x) = (x^2+1)/(x+1)")
for cp in critical_points1:
    if cp.is_real:
        f_val = f1.subs(x, cp)
        concavity = second_derivative1.subs(x, cp)
        if concavity > 0:
            print(f"  Minimum at x = {cp.evalf():.3f}, f(x) = {f_val.evalf():.3f}")
        elif concavity < 0:
            print(f"  Maximum at x = {cp.evalf():.3f}, f(x) = {f_val.evalf():.3f}")
        else:
            print(f"  Saddle point at x = {cp.evalf():.3f}, f(x) = {f_val.evalf():.3f}")

# Plot
x_vals = np.linspace(-5, 5, 400)
f1_np = sp.lambdify(x, f1, "numpy")
y_vals1 = f1_np(x_vals)

plt.figure(figsize=(8,5))
plt.plot(x_vals, y_vals1, label="f1(x) = (x^2+1)/(x+1)")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.ylim(-10, 10)
plt.legend()
plt.grid(True)
plt.show()
```
